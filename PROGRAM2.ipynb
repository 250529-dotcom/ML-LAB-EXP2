```python
import pandas as pd
```

```python
df=pd.read_csv(r"C:\Users\R3ai\Downloads\heart (1).csv")
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np
```

```python
x=df["age"].values.reshape(-1,1)
```

```python
y=df["thalach"]
```

```python
model=LinearRegression()
```

```python
model.fit(x,y)
```

```python
a1=model.coef_[0]
```

```python
a0=model.intercept_
```

```python
print("a1(Slope):",round(a1,2))
```

```python
print("a0 (Interscept)",round(a0,2))
```

```python
from sklearn.metrics import mean_squared_error
```

```python
y_pred_simple=model.predict(x)
```

```python
mse_simple=mean_squared_error(y,y_pred_simple)
```

```python
print("Mean Squared Error (MSE)= ",round(mse_simple,2))
```

```python
import matplotlib.pyplot as plt
plt.scatter(x, y, color="blue")
plt.plot(x, y_pred_simple, color="red")
plt.xlabel("Age")
plt.ylabel("Thalach")
plt.title("Age vs Maximum Heart Rate")
plt.show()
```

Predict thalach using age, trestbps, chol as the independent variables and thalach as the dependent variable
1. train the multilinear regression model
2. find and display coefficients and interscepts, MSE
3. Do Preprocessing Also.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

df = df.drop_duplicates()
df = df.dropna()

X = df[["age", "trestbps", "chol"]]
y = df["thalach"]

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)

print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
print("MSE:", mean_squared_error(y, y_pred))
```
